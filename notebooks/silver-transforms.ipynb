# Silver Layer Transforms — ZOSA Lakehouse
Clean, deduplicate, and type-cast Bronze tables.

Each Bronze table is read, cleaned (nulls removed, strings trimmed, types cast),
deduplicated, and written as a Silver Delta table.

```python
from pyspark.sql.functions import col, to_date, to_timestamp, trim, when, round as spark_round
```

## Asteroids

```python
df = spark.table("asteroids_bronze").dropDuplicates()
df = df.withColumn("name", trim(col("name"))) \
       .withColumn("diameter_km", col("diameter_km").cast("double")) \
       .withColumn("velocity_km_s", col("velocity_km_s").cast("double")) \
       .withColumn("miss_distance_au", col("miss_distance_au").cast("double")) \
       .withColumn("is_hazardous", col("is_hazardous").cast("boolean")) \
       .withColumn("observation_date", to_date(col("observation_date"))) \
       .filter(col("name").isNotNull())
df.write.mode("overwrite").format("delta").saveAsTable("asteroids_silver")
print(f"✅ asteroids_silver: {df.count()} rows")
```

## Solar Events

```python
df = spark.table("solar_events_bronze").dropDuplicates()
df = df.withColumn("event_type", trim(col("event_type"))) \
       .withColumn("intensity", col("intensity").cast("double")) \
       .withColumn("event_date", to_date(col("event_date"))) \
       .withColumn("duration_hours", col("duration_hours").cast("double")) \
       .filter(col("event_type").isNotNull())
df.write.mode("overwrite").format("delta").saveAsTable("solar_events_silver")
print(f"✅ solar_events_silver: {df.count()} rows")
```

## Exoplanets

```python
df = spark.table("exoplanets_bronze").dropDuplicates()
df = df.withColumn("planet_name", trim(col("planet_name"))) \
       .withColumn("star_name", trim(col("star_name"))) \
       .withColumn("orbital_period_days", col("orbital_period_days").cast("double")) \
       .withColumn("radius_earth", col("radius_earth").cast("double")) \
       .withColumn("mass_earth", col("mass_earth").cast("double")) \
       .withColumn("equilibrium_temp_k", col("equilibrium_temp_k").cast("double")) \
       .withColumn("discovery_date", to_date(col("discovery_date"))) \
       .filter(col("planet_name").isNotNull())
df.write.mode("overwrite").format("delta").saveAsTable("exoplanets_silver")
print(f"✅ exoplanets_silver: {df.count()} rows")
```

## Missions

```python
df = spark.table("missions_bronze").dropDuplicates()
df = df.withColumn("mission_name", trim(col("mission_name"))) \
       .withColumn("mission_type", trim(col("mission_type"))) \
       .withColumn("status", trim(col("status"))) \
       .withColumn("launch_date", to_date(col("launch_date"))) \
       .withColumn("end_date", to_date(col("end_date"))) \
       .withColumn("budget_millions", col("budget_millions").cast("double")) \
       .withColumn("region", trim(col("region"))) \
       .filter(col("mission_name").isNotNull())
df.write.mode("overwrite").format("delta").saveAsTable("missions_silver")
print(f"✅ missions_silver: {df.count()} rows")
```

## Crew

```python
df = spark.table("crew_bronze").dropDuplicates()
df = df.withColumn("crew_name", trim(col("crew_name"))) \
       .withColumn("role", trim(col("role"))) \
       .withColumn("nationality", trim(col("nationality"))) \
       .withColumn("experience_years", col("experience_years").cast("int")) \
       .filter(col("crew_name").isNotNull())
df.write.mode("overwrite").format("delta").saveAsTable("crew_silver")
print(f"✅ crew_silver: {df.count()} rows")
```

## Telemetry

```python
df = spark.table("telemetry_bronze").dropDuplicates()
df = df.withColumn("timestamp", to_timestamp(col("timestamp"))) \
       .withColumn("sensor_type", trim(col("sensor_type"))) \
       .withColumn("value", col("value").cast("double")) \
       .withColumn("unit", trim(col("unit"))) \
       .filter(col("timestamp").isNotNull() & col("value").isNotNull())
df.write.mode("overwrite").format("delta").saveAsTable("telemetry_silver")
print(f"✅ telemetry_silver: {df.count()} rows")
```
