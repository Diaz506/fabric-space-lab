# Bronze Layer Ingestion — ZOSA Lakehouse
Load raw CSV files from the Files area into managed Delta tables.

This notebook reads each source CSV and persists it as-is into the Bronze layer,
preserving the original schema. No transformations are applied at this stage.

```python
# --- Asteroids ---
df = spark.read.option("header", True).option("inferSchema", True).csv("Files/asteroids.csv")
df.write.mode("overwrite").format("delta").saveAsTable("asteroids_bronze")
print(f"✅ asteroids_bronze: {df.count()} rows")
```
